```python
import torch
import torch.nn as nn

from easytorch.data import DataLoader
from easytorch.inducer import Inducer
from easytorch.timer import Timer

from example.cifar10 import CIFAR10
from example.module import NeuralNetwork, CorrectRate

if __name__ == "__main__":
  # define variables
  device = "cuda"
  device_ids = [0]
  batch_size = 100
  num_workers = 1
  num_epochs = 10
  learning_rate = 0.005

  # create data set
  cifar10_train = CIFAR10("./data/cifar10/data_batch")
  cifar10_test = CIFAR10("./data/cifar10/test_batch")

  # create data loader
  data_loader = DataLoader(train_data=cifar10_train, test_data=cifar10_test, batch_size=batch_size, num_workers=num_workers)

  # define user-defined network
  model_kwargs = {"in_units" : 3*32*32, "mid_units" : [100, 100, 100, 100], "out_units" : 10}
  model = NeuralNetwork(**model_kwargs)

  # define user-defined criterion
  criterion = nn.CrossEntropyLoss()

  # define user-defined optimizer
  optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)

  # define user-defined evaluator
  evaluator = CorrectRate()

  # define inducer
  inducer = Inducer(model=model, criterion=criterion, evaluator=evaluator, optimizer=optimizer,
                    data_loader=data_loader, save_checkpoint=True,
                    device=device, device_ids=device_ids, non_blocking=True)

  # training
  with Timer() as timer:
    for _ in range(num_epochs):
      inducer.train()
  
  # test
  inducer.test()
```
